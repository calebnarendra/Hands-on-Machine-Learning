# Chapter 11: Training Deep Neural Networks (DNN)

Melatih Deep Neural Networks (DNN) memiliki tantangan tersendiri, terutama terkait stabilitas gradien, optimisasi, dan regularisasi. Chapter ini membahas teknik-teknik penting untuk memastikan proses training berjalan efektif dan efisien.

---

## 11.1 Gradien yang Hilang dan Meledak
Pada proses backpropagation, gradien error cenderung:
- **Menghilang (vanishing gradient)** saat bergerak ke lapisan awal
- **Meledak (exploding gradient)** ketika gradien menjadi sangat besar

Masalah ini menyebabkan:
- Lapisan awal sulit dilatih
- Proses training menjadi tidak stabil atau gagal konvergen

---

## 11.2 Inisialisasi Glorot dan He
Agar sinyal tetap stabil saat propagasi maju dan mundur, variansi input dan output harus seimbang.

### Inisialisasi Glorot (Xavier)
- Bobot diinisialisasi secara acak
- Cocok untuk activation function seperti sigmoid dan tanh
- Berdasarkan jumlah neuron input dan output (fan-in dan fan-out)

### Inisialisasi He
- Dirancang khusus untuk **ReLU**
- Menggunakan skala variansi yang lebih besar
- Lebih efektif menjaga stabilitas gradien pada DNN modern

---

## 11.3 Nonsaturating Activation Functions
Activation function yang tidak mudah jenuh (saturating) membantu mencegah vanishing gradient.

### ReLU
- Sederhana dan efisien
- Namun dapat mengalami **dead neurons** (output selalu nol)
- Biasanya disebabkan learning rate yang terlalu tinggi

### Variasi ReLU
- **Leaky ReLU**: Memiliki gradien kecil untuk nilai negatif
- **Randomized Leaky ReLU (RReLU)**: Parameter bocor dipilih secara acak
- **Parametric ReLU (PReLU)**: Parameter bocor dipelajari dengan backpropagation

---

## 11.4 Normalisasi Batch (Batch Normalization)
Batch Normalization (BN) menambahkan operasi normalisasi pada hidden layer sebelum atau sesudah activation function.

Manfaat:
- Mengurangi vanishing dan exploding gradients
- Mempercepat konvergensi training
- Bertindak sebagai regularisasi ringan

Parameter penting:
- **axis**: Menentukan dimensi yang dinormalisasi menggunakan mean dan standar deviasi

---

## 11.5 Gradient Clipping
Gradient clipping membatasi nilai gradien agar tidak melebihi threshold tertentu.

Tujuan:
- Mencegah exploding gradients
- Menjaga stabilitas proses optimisasi

---

## 11.6 Menggunakan Kembali Pretrained Layers
Melatih DNN dari nol sangat mahal secara komputasi.

**Transfer Learning**:
- Menggunakan model atau layer yang telah dilatih sebelumnya
- Layer awal biasanya menangkap fitur umum
- Sangat efektif untuk dataset kecil

---

## 11.7 Unsupervised Pretraining
Jika tersedia banyak data tanpa label:
- Model dapat dilatih secara **unsupervised**
- Hasilnya digunakan sebagai inisialisasi sebelum supervised learning
- Membantu jaringan mempelajari representasi fitur dasar

---

## 11.8 Pretraining pada Auxiliary Task
Neural network dilatih terlebih dahulu untuk tugas tambahan (auxiliary task).

Manfaat:
- Layer awal belajar mendeteksi fitur umum
- Layer tersebut dapat digunakan kembali untuk tugas utama
- Meningkatkan generalisasi model

---

## 11.9 Optimisasi Momentum
Momentum memperhitungkan gradien dari iterasi sebelumnya.

Karakteristik:
- Mengurangi osilasi gradien
- Mempercepat konvergensi
- Menggunakan **momentum vector** untuk memperbarui bobot

---

## 11.10 Nesterov Accelerated Gradient (NAG)
NAG menghitung gradien sedikit ke depan mengikuti arah momentum.

Keunggulan:
- Prediksi arah minimum lebih akurat
- Konvergensi lebih cepat dibanding momentum standar

---

## 11.11 AdaGrad
AdaGrad menyesuaikan learning rate untuk setiap parameter.

Karakteristik:
- Learning rate menurun pada dimensi dengan gradien besar
- Cocok untuk data sparse
- Namun learning rate bisa menjadi terlalu kecil

## 11.12 RMSProp
RMSProp memperbaiki kelemahan AdaGrad dengan:
- Mengakumulasi gradien dari iterasi terbaru
- Menjaga learning rate tetap stabil di sekitar minimum global

## 11.13 Optimisasi Adam dan Nadam
Adam menggabungkan:
- Momentum
- RMSProp

Varian Adam:
- **AdaMax**: Menggunakan norma maksimum
- **Nadam**: Menambahkan konsep Nesterov ke Adam

Adam adalah optimizer paling populer untuk DNN.

---

## 11.14 Menjadwalkan Learning Rate
Strategi penyesuaian learning rate selama training:
- **Power scheduling**: Menggunakan fungsi pangkat
- **Exponential scheduling**: Learning rate menurun secara eksponensial
- **Piecewise constant scheduling**: Learning rate tetap untuk beberapa epoch
- **Performance scheduling**: Berdasarkan validasi error
- **One-cycle scheduling**: Learning rate naik lalu turun secara linear

---

## 11.15 Regularisasi ℓ1 dan ℓ2
- **ℓ2 (Weight Decay)**: Menahan bobot agar tetap kecil
- **ℓ1**: Menghasilkan model yang lebih sparse

Keduanya membantu mencegah overfitting.

---

## 11.16 Dropout
Dropout mengabaikan neuron secara acak pada setiap step training.

Karakteristik:
- Dropout rate umum: 10% – 50%
- Memaksa jaringan belajar representasi yang lebih robust

### MC Dropout
- Mengaktifkan dropout saat inference
- Mengambil rata-rata beberapa prediksi
- Meningkatkan performa tanpa mengubah arsitektur model

---

## 11.17 Regularisasi Max-Norm
Max-norm membatasi norma bobot pada setiap neuron.

Tujuan:
- Mencegah bobot tumbuh terlalu besar
- Meningkatkan stabilitas dan generalisasi model
