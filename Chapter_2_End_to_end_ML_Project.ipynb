# Chapter 2: End-to-End Machine Learning Project

Dalam proyek Machine Learning, penting untuk bereksperimen menggunakan dataset yang merepresentasikan kondisi dunia nyata. Dataset dapat diperoleh dari berbagai sumber terbuka seperti Kaggle Datasets, AWS Open Datasets, dan Data Portals.  
Contoh kasus yang sering digunakan adalah **memprediksi harga rumah di California**.

---

## 2.1 Menentukan Masalah
Langkah awal adalah mendefinisikan masalah yang ingin diselesaikan. Hal ini mencakup pemahaman bagaimana model akan digunakan setelah dilatih, sehingga dapat menentukan jenis model, parameter, serta *pipeline* yang tepat. Pada tahap ini juga dirancang alur kerja (*workflow*) dari program yang akan dibuat.

---

## 2.2 Memilih Pengukur Performa
Pengukur performa digunakan untuk menilai seberapa baik model dalam melakukan prediksi. Salah satu metrik yang umum digunakan adalah **Root Mean Square Error (RMSE)**, yang mengukur rata-rata kesalahan prediksi model.

**RMSE** mengukur seberapa besar selisih antara nilai prediksi dan nilai sebenarnya.

> *Keterangan:*  
> - **m** : jumlah data  
> - **x** : fitur yang digunakan  
> - **y** : label atau output yang diharapkan  

---

## 2.3 Memastikan Asumsi
Sebelum melanjutkan, perlu dipastikan bahwa pemahaman terhadap data yang akan diolah sudah jelas, baik dari segi struktur, makna setiap atribut, maupun keterkaitannya dengan permasalahan yang ingin diselesaikan.

---

## 2.4 Mengambil Data
Dataset dapat diperoleh dari sumber terpercaya. Salah satu contoh dataset yang sering digunakan tersedia pada repositori berikut:  
https://github.com/ageron/handson-ml2

---

## 2.5 Menciptakan Workspace
Workspace disiapkan menggunakan Python dengan berbagai pustaka pendukung seperti Jupyter Notebook, NumPy, Pandas, Matplotlib, dan Scikit-Learn.  
Direkomendasikan untuk menggunakan *isolated environment* agar proses pelatihan tidak mempengaruhi proyek lain atau sistem utama.

---

## 2.6 Mengunduh Data
Pastikan dataset yang digunakan bersifat terbuka untuk publik dan telah memperoleh izin penggunaan. Data kemudian diunduh dan dimuat melalui platform pemrograman yang digunakan.

---

## 2.7 Melihat Sekilas Struktur Data
Pada tahap ini, struktur dataset dianalisis untuk mengidentifikasi atribut yang tersedia, seperti jumlah kamar, populasi, atau pendapatan. Atribut-atribut tersebut dievaluasi untuk menentukan mana yang relevan dalam proses prediksi. Data juga dapat divisualisasikan menggunakan Matplotlib.

---

## 2.8 Menciptakan Test Set
Sebelum model diluncurkan, sebagian kecil dataset perlu disisihkan sebagai *test set*. Test set digunakan untuk mengevaluasi performa model secara objektif terhadap data yang belum pernah dilihat sebelumnya.

---

## 2.9 Memvisualisasikan Data Geografis
Untuk memperoleh gambaran lebih baik mengenai distribusi data, visualisasi geografis dapat dilakukan. Dalam kasus prediksi harga rumah, visualisasi dapat difokuskan pada area dengan kepadatan penduduk yang tinggi.

---

## 2.10 Mencari Korelasi
Pada tahap ini dianalisis atribut mana yang memiliki pengaruh paling besar terhadap label yang ingin diprediksi. Korelasi antar atribut dihitung menggunakan parameter statistik tertentu.

---

## 2.11 Bereksperimen dengan Kombinasi Atribut
Berbagai kombinasi atribut diuji untuk memastikan model dapat mempelajari hubungan yang relevan dan meningkatkan kualitas prediksi.

---

## 2.12 Menyiapkan Data
Setelah evaluasi awal selesai, data dipersiapkan secara menyeluruh untuk dimasukkan ke dalam proses Machine Learning.

---

## 2.13 Membersihkan Data
Dataset sering kali memiliki nilai yang hilang atau tidak lengkap. Masalah ini dapat diatasi dengan menghapus data yang tidak valid atau menggantinya dengan nilai tertentu, seperti rata-rata atau median.

---

## 2.14 Mengolah Atribut Non-Angka
Beberapa atribut bersifat kategorikal dan tidak dapat langsung diproses oleh model. Atribut ini perlu dikonversi menjadi bentuk numerik, salah satunya menggunakan **Ordinal Encoder** dari Scikit-Learn.

---

## 2.15 Custom Transformers
Transformers khusus dapat dibuat untuk menangani proses data yang lebih kompleks. Custom transformer diimplementasikan menggunakan metode `fit()`, `transform()`, dan `fit_transform()`.

---

## 2.16 Mengatur Skala Fitur
Ketika fitur memiliki skala yang sangat berbeda (misalnya 0–15 pada satu fitur dan 6–39.320 pada fitur lain), diperlukan normalisasi atau standarisasi agar model dapat memproses data secara lebih efektif.

---

## 2.17 Perubahan Pipeline
Scikit-Learn menyediakan *pipeline* untuk menggabungkan seluruh tahapan pemrosesan data dan pelatihan model, sehingga alur kerja menjadi lebih terstruktur dan konsisten.

---

## 2.18 Training dan Evaluating Training Set
Dataset pelatihan digunakan untuk melatih model, seperti Linear Regression, dan kemudian dievaluasi. Jika nilai error terlalu besar atau terlalu kecil, metode evaluasi perlu disesuaikan, salah satunya dengan menggunakan *cross-validation*.

---

## 2.19 Grid Search
Grid Search digunakan untuk melakukan *hyperparameter tuning* pada model yang dipilih dengan mengevaluasi seluruh kombinasi parameter menggunakan **GridSearchCV**.

---

## 2.20 Randomized Search
Ketika ruang hyperparameter sangat luas, **RandomizedSearchCV** digunakan untuk memilih kombinasi parameter secara acak pada setiap iterasi, sehingga lebih efisien secara komputasi.

---

## 2.21 Metode Ensemble
Metode ensemble menggabungkan beberapa model dengan performa baik untuk menghasilkan prediksi yang lebih stabil dan akurat.

---

## 2.22 Analisis Error dan Peluncuran Sistem
Tahap akhir meliputi analisis error dari setiap model, evaluasi sistem menggunakan *test set*, serta peluncuran sistem ke lingkungan nyata.
