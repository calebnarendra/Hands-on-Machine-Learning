# Chapter 8: Dimensionality Reduction

Dimensionality Reduction bertujuan untuk mengurangi jumlah fitur dalam dataset sambil mempertahankan informasi penting, sehingga proses training menjadi lebih efisien dan stabil.

---

## 8.1 Curse of Dimensionality
Setiap dimensi dalam dataset memiliki karakteristik yang berbeda.

Semakin tinggi dimensi suatu dataset, semakin jarang data tersebar secara efektif di ruang fitur, sehingga proses training menjadi lebih sulit dan hasil prediksi di dunia nyata dapat menjadi tidak akurat.

---

## 8.2 Projection
Pada data dunia nyata, contoh training biasanya tidak tersebar merata pada seluruh dimensi.

Sebagian besar data berada pada subruang berdimensi lebih rendah, sehingga proyeksi ke dimensi yang lebih rendah dapat menangkap struktur data dengan lebih tepat.

---

## 8.3 Manifold Learning
Manifold Learning berasumsi bahwa data berdimensi tinggi sebenarnya terletak pada manifold berdimensi lebih rendah.

Contoh klasiknya adalah **Swiss Roll**, yaitu manifold 2D yang terlipat dalam ruang berdimensi lebih tinggi.  
Pada dataset seperti MNIST, data dengan kemiripan tinggi cenderung berada pada manifold yang sama, sehingga prediksi menjadi lebih akurat dibandingkan menggunakan data acak.

---

## 8.4 Menjaga Variansi pada PCA
Dalam **Principal Component Analysis (PCA)**, pemilihan bidang proyeksi yang tepat sangat penting.

Bidang yang dipilih harus mempertahankan variansi terbesar, karena variansi tinggi mengandung informasi yang paling relevan dari data.

---

## 8.5 Principal Components
Principal Components merupakan sumbu-sumbu baru yang menyimpan informasi paling besar dari data.

Semakin tinggi dimensi awal, semakin banyak principal components yang tersedia. Komponen-komponen ini umumnya dihitung menggunakan **Singular Value Decomposition (SVD)**.

---

## 8.6 Menurunkan Proyeksi
Dimensi data dapat diturunkan dengan memilih sejumlah principal components teratas dan memproyeksikan data ke bidang berdimensi lebih rendah yang telah ditentukan.

---

## 8.7 Rasio Variansi
Setiap principal component memiliki **explained variance ratio**, yang menunjukkan proporsi variansi data yang dijelaskan oleh komponen tersebut.

---

## 8.8 Memilih Dimensi yang Tepat
Jumlah dimensi yang optimal dipilih dengan menjumlahkan explained variance hingga mencapai nilai variansi kumulatif yang diinginkan.

Dimensi yang menghasilkan variansi terbesar dianggap paling representatif untuk data.

---

## 8.9 Kompresi dengan PCA
Dengan mempertahankan variansi yang tinggi, PCA dapat mengurangi jumlah fitur sehingga data membutuhkan ruang penyimpanan yang lebih kecil.

Data hasil kompresi juga dapat direkonstruksi kembali, dan kualitasnya dievaluasi menggunakan **reconstruction error**.

---

## 8.10 PCA yang Diacak
**Randomized PCA** memiliki kompleksitas komputasi yang lebih rendah dibandingkan SVD penuh, yaitu sekitar  
**O(m × d²) + O(d³)**, sehingga lebih efisien untuk dataset besar.

---

## 8.11 Incremental PCA
Salah satu keterbatasan PCA adalah kebutuhan untuk memuat seluruh training set ke dalam memori.

**Incremental PCA** mengatasi masalah ini dengan memproses data secara bertahap dalam beberapa bagian kecil.

---

## 8.12 Kernel PCA
Kernel PCA memungkinkan PCA diterapkan pada proyeksi non-linear yang kompleks.

Pemilihan kernel dan parameter yang tepat dilakukan dengan mengevaluasi hasil reduksi dimensi, salah satunya melalui minimisasi error rekonstruksi.

---

## 8.13 Locally Linear Embedding (LLE)
**Locally Linear Embedding (LLE)** merupakan teknik manifold learning untuk mereduksi dimensi data non-linear tanpa bergantung pada proyeksi global.

Pada setiap tahap training, algoritma mengidentifikasi **k tetangga terdekat** untuk merekonstruksi setiap contoh sebagai kombinasi linear dari tetangganya.  
Hasil rekonstruksi kemudian diproyeksikan ke ruang berdimensi lebih rendah dengan tetap menjaga hubungan lokal antar data.
