# Chapter 7: Ensemble Learning and Random Forests

Ensemble Learning merupakan pendekatan Machine Learning yang menggabungkan prediksi dari beberapa model untuk menghasilkan performa yang lebih baik dibandingkan satu model tunggal.

---

## 7.1 Ensemble Learning
Ensemble Learning menggabungkan prediksi dari berbagai model, baik classifier maupun regressor, untuk meningkatkan akurasi dan stabilitas prediksi.

Pada Decision Tree, kumpulan model ensemble ini dikenal sebagai **Random Forest**.

---

## 7.2 Voting Classifiers
Voting classifier menggabungkan beberapa classifier dan menentukan kelas berdasarkan hasil voting.

- **Hard voting**: kelas yang memperoleh suara terbanyak dipilih sebagai hasil prediksi  
- **Soft voting**: menggunakan rata-rata probabilitas prediksi dari setiap model  

Pendekatan ini bekerja efektif berkat **law of large numbers**, di mana kesalahan model individu dapat saling menutupi.

---

## 7.3 Bagging and Pasting
Bagging dan pasting melatih beberapa model menggunakan subset berbeda dari *training set* yang sama.

- **Bagging (Bootstrap Aggregating)**: sampling dilakukan dengan pengembalian (*with replacement*)  
- **Pasting**: sampling dilakukan tanpa pengembalian  

Pada Scikit-Learn, metode ini diimplementasikan menggunakan **BaggingClassifier**.  
Secara umum, bagging cenderung menurunkan variansi model dibandingkan pasting.

---

## 7.4 Out-of-Bag Evaluation
Pada metode bagging, sebagian data pelatihan tidak terambil dalam proses sampling.

Data tersebut disebut **out-of-bag (OOB)** dan dapat digunakan sebagai data validasi untuk mengevaluasi performa model tanpa memerlukan *validation set* terpisah.

---

## 7.5 Random Patches dan Random Subspaces
Pada BaggingClassifier, teknik sampling dapat diperluas dengan:
- **Random patches**: sampling dilakukan pada data dan fitur sekaligus  
- **Random subspaces**: sampling hanya dilakukan pada fitur  

Pendekatan ini dikendalikan melalui hyperparameter **max_features** dan **bootstrap_features**, dan sangat berguna untuk data berdimensi tinggi.

---

## 7.6 Extra-Trees
**Extremely Randomized Trees (Extra-Trees)** meningkatkan tingkat keacakan dengan memilih threshold secara acak untuk setiap fitur, alih-alih mencari threshold terbaik.

Pendekatan ini mempercepat training dan mengurangi variansi dengan sedikit peningkatan bias.

---

## 7.7 Kepentingan Fitur
Random Forest dan Extra-Trees dapat mengukur **feature importance**.

Kepentingan suatu fitur ditentukan berdasarkan seberapa besar fitur tersebut berkontribusi dalam mengurangi ketidakmurnian (*impurity*) data secara rata-rata di seluruh pohon.

---

## 7.8 Boosting
Boosting merupakan teknik ensemble yang menggabungkan beberapa model lemah (*weak learners*) secara sekuensial untuk membentuk model yang kuat.

---

## 7.9 AdaBoost
**Adaptive Boosting (AdaBoost)** memberikan perhatian lebih besar pada contoh training yang sebelumnya sulit diprediksi.

Setelah model awal dilatih, kesalahan rata-rata dihitung, lalu bobot pada contoh training dan model prediksi disesuaikan dan dinormalisasi agar model berikutnya fokus pada kasus yang lebih sulit.

---

## 7.10 Gradient Boosting
Gradient Boosting bekerja secara sekuensial, di mana setiap model baru dilatih untuk memperbaiki kesalahan yang dibuat oleh model sebelumnya.

Pendekatan ini memanfaatkan optimasi berbasis gradien untuk meminimalkan fungsi biaya secara bertahap.

---

## 7.11 Stacking
Stacking menggabungkan prediksi dari berbagai model dengan melatih sebuah model tambahan.

Model terakhir ini disebut **blender**, yang dilatih menggunakan *hold-out set* untuk mempelajari cara terbaik menggabungkan prediksi dari seluruh model sebelumnya.
