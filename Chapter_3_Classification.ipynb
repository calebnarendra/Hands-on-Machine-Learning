# Chapter 3: Classification

Klasifikasi merupakan salah satu tugas utama dalam Machine Learning yang bertujuan untuk memetakan input ke dalam kategori atau kelas tertentu.

---

## 3.1 Dataset MNIST
MNIST adalah dataset yang terdiri dari **70.000 gambar angka tulisan tangan** berukuran kecil (28×28 piksel), yang ditulis oleh siswa dan karyawan Biro Sensus Amerika Serikat.  
Setiap gambar telah diberi label angka (0–9) yang merepresentasikan digit sebenarnya. Dataset ini umum digunakan untuk melatih dan menguji model klasifikasi angka tulisan tangan.

---

## 3.2 Melatih Binary Classifier
Sebagai penyederhanaan awal, masalah klasifikasi dapat dimulai dengan **binary classification**, yaitu membedakan satu digit tertentu dari digit lainnya.  
Contohnya, mengklasifikasikan apakah sebuah gambar merupakan digit *5* atau bukan *5*.

---

## 3.3 Mengukur Performa dengan Cross-Validation
Performa model dievaluasi menggunakan **K-fold cross-validation**, khususnya dengan **StratifiedKFold**, agar proporsi kelas tetap seimbang di setiap fold.  

Dalam kasus dataset yang tidak seimbang, **akurasi** bukanlah metrik performa yang ideal. Model dapat memperoleh akurasi tinggi hanya dengan menebak kelas mayoritas, meskipun gagal mengenali kelas target yang sebenarnya.

---

## 3.4 Confusion Matrix
Confusion Matrix digunakan untuk mengevaluasi hasil prediksi model dengan membandingkan nilai aktual dan nilai prediksi.  
Matriks ini membagi hasil klasifikasi ke dalam empat kategori:
- **True Positive (TP)**
- **True Negative (TN)**
- **False Positive (FP)**
- **False Negative (FN)**

Evaluasi dilakukan menggunakan data yang bersih, karena data yang dianalisis merupakan hasil prediksi model pada setiap iterasi.

---

## 3.5 Precision dan Recall
Precision dan recall merupakan metrik yang diturunkan dari Confusion Matrix.

- **Precision** mengukur seberapa akurat prediksi positif yang dihasilkan model  
- **Recall** mengukur seberapa banyak data positif yang berhasil dikenali oleh model  

Dalam klasifikasi, dikenal konsep **precision–recall trade-off**, yaitu peningkatan precision sering kali menurunkan recall, dan sebaliknya. Oleh karena itu, perlu ditentukan metrik mana yang lebih penting sesuai dengan kebutuhan sistem.

## 3.6 Kurva ROC
**Receiver Operating Characteristic (ROC)** merupakan metode evaluasi yang digunakan pada binary classifier.  
Kurva ROC menampilkan hubungan antara:
- **True Positive Rate (TPR)**
- **False Positive Rate (FPR)**  

Kurva ini membantu dalam mengevaluasi performa model pada berbagai nilai ambang (*threshold*).

## 3.7 Klasifikasi Multiclass
Klasifikasi multiclass merupakan pengembangan dari binary classification, di mana model harus membedakan lebih dari dua kelas.  
Algoritma seperti **Stochastic Gradient Descent (SGD)**, **Random Forest**, dan **Naïve Bayes** dapat digunakan untuk mengklasifikasikan dataset dengan banyak kelas, seperti pengenalan digit 0 hingga 9 pada MNIST.

---

## 3.8 Klasifikasi Multilabel
Klasifikasi multilabel memungkinkan satu data memiliki lebih dari satu label secara bersamaan.  
Pendekatan ini dapat diterapkan menggunakan algoritma seperti **KNeighborsClassifier**, yang mampu menangani beberapa label pada satu instance data.

---

## 3.9 Klasifikasi Multioutput
Klasifikasi multioutput merupakan generalisasi dari klasifikasi multilabel, di mana setiap label dapat memiliki lebih dari dua kelas (*multiclass*).  
Pendekatan ini digunakan pada sistem yang menghasilkan beberapa output klasifikasi secara bersamaan.
