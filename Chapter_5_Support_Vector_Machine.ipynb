# Chapter 5: Support Vector Machines

Support Vector Machines (SVM) merupakan algoritma Machine Learning yang kuat untuk tugas klasifikasi dan regresi, terutama ketika terdapat margin pemisah yang jelas antar kelas.

---

## 5.1 Klasifikasi SVM Linear
SVM linear dikenal sebagai metode klasifikasi dengan **margin maksimum**, yang bertujuan memisahkan kelas-kelas data dengan batas keputusan yang sejauh mungkin dari data terdekat.

Pendekatan ini efektif ketika data dapat dipisahkan secara linear.

---

## 5.2 Soft Margin Classification
Klasifikasi SVM linear ideal hanya berlaku jika data benar-benar dapat dipisahkan secara sempurna. Namun, pada data nyata sering terdapat *outlier*.

Soft margin classification memberikan fleksibilitas dengan mengizinkan sebagian data berada di dalam margin atau salah klasifikasi. Parameter **hyperparameter** (seperti *C*) mengatur keseimbangan antara lebar margin dan jumlah kesalahan klasifikasi.

---

## 5.3 Klasifikasi SVM Non-linear
Untuk data yang tidak dapat dipisahkan secara linear, diperlukan pendekatan non-linear. Hal ini dapat dilakukan dengan menambahkan fitur baru atau menggunakan teknik kernel agar data dapat dipisahkan pada ruang berdimensi lebih tinggi.

---

## 5.4 Polynomial Kernel
Polynomial kernel menggunakan kombinasi fitur polynomial berorde lebih tinggi untuk memodelkan hubungan yang kompleks.

Melalui *kernel trick*, SVM dapat menghasilkan hasil yang setara dengan penggunaan fitur berdimensi tinggi tanpa harus menghitung fitur tersebut secara eksplisit.

---

## 5.5 Kemiripan Fitur
Pendekatan kemiripan fitur dapat dilakukan menggunakan **Gaussian Radial Basis Function (RBF)**.  
Metode ini bekerja dengan menentukan sejumlah *landmark* pada dataset dan mengukur tingkat kemiripan setiap data terhadap landmark tersebut.

Gaussian RBF juga dapat diterapkan dalam bentuk kernel, termasuk pada kasus khusus seperti *string kernels* untuk klasifikasi data sekuensial, misalnya DNA.

## 5.6 Kompleksitas Komputasi
Kompleksitas komputasi SVM bergantung pada jumlah data (**m**) dan jumlah fitur (**n**):

- SVM linear memiliki kompleksitas training sekitar **O(m × n)**  
- SVM dengan kernel memiliki kompleksitas sekitar **O(m² × n)** hingga **O(m³ × n)**  

Akibatnya, SVM berbasis kernel menjadi semakin lambat ketika jumlah data pelatihan meningkat.

---

## 5.7 Regresi SVM
SVM juga dapat digunakan untuk regresi (*Support Vector Regression*).  
Tujuannya adalah menemukan fungsi yang menyimpang seminimal mungkin dari data aktual dengan toleransi kesalahan tertentu.

Dalam SVR, hanya sebagian data (support vectors) yang mempengaruhi prediksi, sehingga penambahan data baru tidak selalu mengubah model.

---

## 5.8 Fungsi Keputusan dan Prediksi
Prediksi pada SVM dilakukan berdasarkan **fungsi keputusan**, yang menentukan posisi suatu data relatif terhadap margin dan hyperplane pemisah.

---

## 5.9 Tujuan Training
Tujuan utama training SVM adalah meminimalkan norma vektor bobot, yang secara langsung akan memperlebar margin.

Pendekatan ini disesuaikan dengan jenis klasifikasi yang digunakan, yaitu **hard margin classifier** atau **soft margin classifier**.

---

## 5.13 Online SVMs
Untuk kasus data yang terus bertambah, SVM dapat dilatih secara *online*.

Pendekatan ini menggunakan **penurunan gradien** pada linear SVM classifiers dengan fungsi biaya  
**max(0, 1 − t)** yang dikenal sebagai **hinge loss**.
