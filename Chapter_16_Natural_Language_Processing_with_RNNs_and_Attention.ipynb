# Chapter 16: Natural Language Processing with RNNs and Attention

Bab ini membahas teknik NLP tingkat lanjut, mulai dari pembuatan teks otomatis, analisis sentimen, hingga arsitektur Encoder-Decoder dengan mekanisme Attention untuk penerjemahan mesin.

---

## 16.1 Menghasilkan Teks (Char-RNN)
Model **Char-RNN** dilatih untuk memprediksi karakter berikutnya dalam sebuah urutan.
* **Proses**: Dataset teks (seperti karya Shakespeare) dipecah menjadi urutan karakter yang lebih kecil menggunakan metode `.window()` agar komputasi lebih ringan.
* **Arsitektur**: Biasanya terdiri dari lapisan **GRU** atau **LSTM** dengan *dropout* untuk mencegah overfitting.
* **Tujuan**: Model belajar menangkap struktur tata bahasa dan gaya penulisan dari data latih.

---

## 16.2 Stateful RNN
Dalam RNN standar (*stateless*), *hidden state* direset menjadi nol setiap kali memproses *batch* baru.
* **Stateful RNN**: Menyimpan *state* terakhir dari satu *batch* dan menggunakannya sebagai *state* awal untuk *batch* berikutnya.
* **Syarat**: Input harus sekuensial (berurutan) dan tidak boleh tumpang tindih (*non-overlapping*). Ini sangat berguna untuk mempelajari pola yang sangat panjang.

---

## 16.3 Sentiment Analysis (IMDb)
Analisis sentimen pada dataset IMDb sering dianggap sebagai "Hello World" dalam NLP.
* **Preprocessing**: Teks mentah dibersihkan dan diubah menjadi angka menggunakan **Tokenizer**.
* **Embedding**: Kata-kata diubah menjadi vektor padat (*dense vectors*) yang mewakili makna semantik kata tersebut.

---

## 16.4 Masking
Karena kalimat dalam satu *batch* memiliki panjang yang berbeda, kita menggunakan *padding* (biasanya angka 0).
* **Mekanisme**: Dengan `mask_zero=True` pada `Embedding layer`, model akan mengabaikan nilai *padding* tersebut selama proses kalkulasi agar tidak merusak hasil prediksi atau perhitungan gradien.

---

## 16.5 Reusing Pretrained Embeddings
Alih-alih melatih *embedding* dari nol, kita bisa menggunakan model yang sudah dilatih pada dataset raksasa (seperti Google News atau Wikipedia).
* **Keuntungan**: Membantu model memahami hubungan antar kata dengan lebih baik, terutama jika dataset lokal kita jumlahnya terbatas.

---

## 16.6 Encoder-Decoder Network
Digunakan dalam *Neural Machine Translation* (NMT).
* **Encoder**: Memproses kalimat bahasa asal menjadi vektor representasi tunggal.
* **Decoder**: Mengambil vektor tersebut dan menghasilkan kalimat bahasa tujuan satu per satu.
* **Tokens**: Menggunakan **SOS** (*Start-of-Sequence*) untuk memulai dan **EOS** (*End-of-Sequence*) untuk menandai akhir kalimat.



---

## 16.7 Bidirectional RNNs
RNN standar hanya melihat informasi dari masa lalu.
* **Cara Kerja**: Lapisan ini memproses sekuens dua kali; satu dari kiri ke kanan (maju) dan satu dari kanan ke kiri (mundur).
* **Hasil**: Output dari kedua arah digabungkan. Ini sangat efektif untuk memahami konteks kata yang bergantung pada kata-kata setelahnya.

---

## 16.8 Beam Search
Saat men-decode kalimat, mengambil satu kata dengan probabilitas tertinggi di setiap langkah (Greedy Search) tidak selalu menghasilkan kalimat terbaik.
* **Beam Search**: Menyimpan daftar pendek (misal 5 kandidat) urutan kata yang paling mungkin pada setiap langkah dan membandingkannya secara keseluruhan untuk mendapatkan hasil akhir yang paling masuk akal.

---

## 16.9 Attention Mechanisms
Mekanisme yang memungkinkan decoder untuk "melihat kembali" ke bagian tertentu dari kalimat input pada setiap langkah prediksi.
* **Manfaat**: Mengatasi keterbatasan memori jangka pendek pada RNN. Decoder tidak lagi hanya bergantung pada satu vektor ringkasan, tapi bisa fokus pada kata yang relevan.
* **Alignment Model**: Sub-network kecil di dalam model yang menentukan bagian input mana yang harus diberi perhatian lebih.



---

## 16.10 Visual Attention & Scaled Dot-Product
* **Visual Attention**: Digunakan untuk deskripsi gambar (*image captioning*). Decoder fokus pada area tertentu dalam gambar saat menghasilkan kata tertentu (misal: fokus ke area "bola" saat menulis kata "bola").
* **Scaled Dot-Product Attention**: Inti dari arsitektur **Transformer**. Menggunakan representasi vektor berupa *Query*, *Key*, dan *Value* untuk menghitung kecocokan dan memberikan bobot pada informasi yang paling relevan tanpa perlu menggunakan RNN secara eksplisit.
