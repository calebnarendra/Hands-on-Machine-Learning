# Chapter 6: Decision Trees

Decision Tree merupakan model Machine Learning yang bekerja dengan membagi data secara berurutan berdasarkan kondisi tertentu hingga mencapai keputusan akhir.

---

## 6.1 Training dan Visualisasi
Proses training Decision Tree dapat divisualisasikan untuk memahami alur pengambilan keputusan model.

Pada Scikit-Learn, visualisasi pohon keputusan dapat dilakukan menggunakan fungsi **export_graphviz()** dari modul `sklearn.tree`.

---

## 6.2 Melakukan Prediksi
Prediksi dilakukan dengan menelusuri pohon dari **root node** menuju **leaf node**.

Setiap node menyimpan informasi mengenai jumlah sampel training yang melewati node tersebut. Model menentukan prediksi dengan mengikuti cabang kiri atau kanan berdasarkan nilai fitur hingga mencapai leaf node.

---

## 6.3 Probabilitas Estimasi Kelas
Decision Tree dapat mengestimasi probabilitas suatu kelas berdasarkan proporsi contoh training dari masing-masing kelas yang berada pada leaf node.

Kelas dengan probabilitas tertinggi akan dipilih sebagai hasil prediksi.

---

## 6.4 CART Training Algorithm
Decision Tree umumnya dilatih menggunakan algoritma **CART (Classification and Regression Trees)**.

Algoritma ini membagi *training set* menjadi dua subset berdasarkan sebuah fitur **k** dan nilai ambang (*threshold*) **t**, kemudian mengevaluasi kualitas split berdasarkan bobot subset yang dihasilkan.

CART menggunakan pendekatan **greedy algorithm**, yaitu memilih split paling optimal pada level saat ini sebelum melanjutkan ke level berikutnya.

---

## 6.5 Kompleksitas Komputasi
Menelusuri sebuah decision tree untuk melakukan prediksi memiliki kompleksitas sekitar **O(log₂(m))**, dengan **m** sebagai jumlah data.

Namun, proses training yang membandingkan seluruh fitur memiliki kompleksitas sekitar **O(n × m log₂(m))**, dengan **n** sebagai jumlah fitur.

---

## 6.6 Entropi
Entropi digunakan untuk mengukur tingkat kemurnian (*purity*) sebuah dataset.

Nilai entropi akan mencapai nol ketika semua data dalam suatu node berasal dari kelas yang sama. Semakin kecil nilai entropi, semakin homogen data dalam node tersebut.

---

## 6.7 Regularization Hyperparameters
Berbeda dengan model linear, Decision Tree tidak membuat banyak asumsi terhadap data pelatihan. Tanpa pengawasan, model ini sangat mudah mengalami **overfitting**.

Decision Tree termasuk model **nonparametric**, karena jumlah parameter tidak ditentukan sebelum training. Oleh karena itu, diperlukan teknik regularisasi dengan membatasi kebebasan model, seperti:
- Membatasi kedalaman pohon (*max depth*)
- Membatasi jumlah sampel minimum pada node

---

## 6.8 Regresi
Selain klasifikasi, Decision Tree juga dapat digunakan untuk regresi.

Dalam regresi, model memprediksi nilai kontinu, dan algoritma CART memisahkan *training set* dengan tujuan meminimalkan **Mean Squared Error (MSE)**.

---

## 6.9 Ketidakstabilan
Decision Tree sangat sensitif terhadap variasi kecil pada data pelatihan.

Perubahan kecil pada training set atau rotasi data dapat menghasilkan struktur pohon yang sangat berbeda, sehingga model cenderung tidak stabil.
