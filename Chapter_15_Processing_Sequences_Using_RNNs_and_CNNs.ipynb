# Chapter 15: Processing Sequences Using RNNs and CNNs

Bab ini membahas cara menangani data sekuensial (deret waktu, teks, atau audio) menggunakan Recurrent Neural Networks (RNN) dan bagaimana 1D Convolutional Layers dapat membantu mempercepat proses tersebut.

---

## 15.1 Neuron dan Lapisan Berulang (Recurrent Layers)
Berbeda dengan *feedforward neural network* yang informasinya mengalir satu arah, RNN memiliki koneksi yang mengarah ke belakang.
* **Mekanisme**: Satu neuron menerima input $x_{(t)}$, menghasilkan output $\hat{y}_{(t)}$, dan mengirimkan output tersebut kembali ke dirinya sendiri sebagai input tambahan untuk langkah waktu berikutnya ($t+1$).
* Jaringan ini dapat dianggap sebagai tumpukan lapisan yang identik jika "dibuka" (*unrolled*) sepanjang sumbu waktu.



---

## 15.2 Sel Memori (Memory Cells)
Karena output dari suatu *time step* bergantung pada input dari langkah sebelumnya, RNN memiliki bentuk memori.
* **Cell State**: Bagian dari RNN yang menjaga kondisi internal di berbagai langkah waktu disebut sel memori.
* Sel ini menyimpan informasi relevan dari masa lalu untuk mempengaruhi prediksi di masa depan.

---

## 15.3 Arsitektur Sekuens Input dan Output
RNN sangat fleksibel dalam menangani berbagai jenis data:
1. **Sequence-to-Sequence**: Input deret waktu, output deret waktu (misal: prediksi harga saham harian).
2. **Sequence-to-Vector**: Input deret waktu, mengabaikan semua output kecuali yang terakhir (misal: analisis sentimen kalimat).
3. **Vector-to-Sequence**: Input tunggal, output deret waktu (misal: *image captioning*).
4. **Encoder-Decoder**: Menggabungkan *sequence-to-vector* (encoder) diikuti oleh *vector-to-sequence* (decoder) untuk terjemahan mesin.

---

## 15.4 Pelatihan RNN (BPTT)
Pelatihan RNN menggunakan teknik **Backpropagation Through Time (BPTT)**.
* Jaringan "dibuka" sepanjang waktu, lalu algoritma backpropagation standar diterapkan.
* Gradien mengalir mundur melalui semua langkah waktu untuk memperbarui bobot.

---

## 15.5 Meramal Deret Waktu (Time Series Forecasting)
* **Univariate**: Hanya ada satu variabel yang diamati per langkah waktu (misal: suhu saja).
* **Multivariate**: Ada beberapa variabel (misal: suhu, kelembapan, dan tekanan udara).
* **Forecasting**: Menebak nilai di masa depan.
* **Imputation**: Mengisi nilai yang hilang di masa lalu.

---

## 15.6 Baseline Metrics & Naïve Forecasting
Sebelum menggunakan model rumit, kita butuh pembanding (*baseline*).
* **Naïve Forecasting**: Metode paling sederhana dengan mengasumsikan nilai di waktu $t+1$ adalah sama dengan nilai di waktu $t$.
* Model RNN yang baik harus mampu mengalahkan performa *naïve forecasting* ini.

---

## 15.7 Masalah Gradien yang Tidak Stabil
RNN sering mengalami *vanishing* atau *exploding gradients* karena perkalian bobot yang berulang kali sepanjang langkah waktu.
* **Solusi**: Menggunakan *Learning Rate* yang kecil, *Gradient Clipping*, atau fungsi aktivasi yang lebih stabil (saturating activation functions).

---

## 15.8 Mengatasi Masalah Memori Jangka Pendek
RNN standar sulit mengingat informasi yang terjadi jauh di masa lalu.
* **LSTM (Long Short-Term Memory)**: Memiliki jalur khusus (*cell state*) yang memungkinkan informasi mengalir tanpa banyak perubahan. Memiliki *forget gate*, *input gate*, dan *output gate*.
* **Peephole Connections**: Variasi LSTM di mana informasi dari *cell state* juga dikirim ke pengatur *gate*.
* **GRU (Gated Recurrent Unit)**: Versi LSTM yang lebih sederhana dengan menggabungkan *forget* dan *input gate*. Lebih cepat dihitung namun tetap efektif.



---

## 15.9 Menggunakan 1D Convolutional Layers
Ternyata, konvolusi 1D dapat digunakan untuk memproses sekuens.
* **Cara Kerja**: *Kernel* bergeser sepanjang sumbu waktu (bukan tinggi/lebar seperti pada gambar).
* **Keuntungan**: Jauh lebih cepat daripada RNN dan efektif untuk menangkap pola lokal dalam deret waktu yang sangat panjang.
