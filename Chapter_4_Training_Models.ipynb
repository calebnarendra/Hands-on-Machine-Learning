# Chapter 4: Training Models

Tahap training model merupakan inti dari proses Machine Learning, di mana model belajar dari data untuk menemukan pola dan hubungan yang optimal.

---

## 4.1 Regresi Linear
Regresi linear merupakan model dasar yang digunakan untuk memodelkan hubungan antara fitur dan target menggunakan persamaan linear.

image.png

Beberapa metrik yang umum digunakan untuk mengevaluasi regresi linear antara lain:
- **Mean Squared Error (MSE)**
- **Root Mean Squared Error (RMSE)**
- **Mean Absolute Error (MAE)**

---

## 4.2 Persamaan Normal
Persamaan normal merupakan solusi tertutup (*closed-form solution*) untuk mencari nilai parameter yang meminimalkan fungsi biaya pada regresi linear.

Untuk menyelesaikan persamaan ini secara numerik, digunakan teknik faktorisasi matriks seperti **Singular Value Decomposition (SVD)**.

image.png

---

## 4.3 Kompleksitas Komputasi
Waktu komputasi ditentukan oleh cara persamaan menghitung fitur-fitur pada matriks data.

- Persamaan normal memiliki kompleksitas sekitar **O(n³)**
- SVD memiliki kompleksitas sekitar **O(n²)**

Semakin banyak fitur yang digunakan, semakin besar pula kompleksitas komputasi yang dibutuhkan.

---

## 4.4 Penurunan Gradien (Gradient Descent)
Gradient Descent merupakan metode optimasi umum yang digunakan untuk mencari solusi optimal dengan cara memperbarui parameter secara iteratif guna meminimalkan fungsi biaya.

Parameter **learning rate** harus diatur dengan tepat. Jika terlalu kecil, proses konvergensi akan sangat lambat; jika terlalu besar, model dapat gagal mencapai nilai minimum.

---

## 4.5 Penurunan Gradien Bertahap (Batch Gradient Descent)
Metode ini menghitung gradien menggunakan seluruh dataset pada setiap iterasi untuk menentukan perubahan fungsi biaya menuju nilai minimum.

---

## 4.6 Penurunan Gradien Stokastik (Stochastic Gradient Descent)
Berbeda dari metode batch, pendekatan ini menghitung gradien menggunakan satu data acak pada setiap langkah.

Metode ini lebih cocok untuk *training set* berukuran besar, tetapi fungsi biaya cenderung berfluktuasi secara ekstrem dan mendekati nilai minimum melalui rata-rata.

---

## 4.7 Penurunan Gradien Mini-batch
Mini-batch Gradient Descent merupakan gabungan dari dua metode sebelumnya, dengan menghitung gradien menggunakan sekumpulan kecil data yang dipilih secara acak pada setiap iterasi.

---

## 4.8 Regresi Polynomial
Untuk data yang bersifat non-linear, regresi polynomial digunakan dengan menambahkan pangkat tertentu pada fitur agar hubungan non-linear dapat dimodelkan.

Pendekatan ini dapat diimplementasikan menggunakan **PolynomialFeatures** pada Scikit-Learn.

---

## 4.9 Kurva Pembelajaran
Selain cross-validation, performa generalisasi model dapat dianalisis melalui **learning curve**.

- **Bias** terjadi ketika model terlalu sederhana dan gagal menangkap pola data  
- **Variansi** terjadi ketika model terlalu sensitif terhadap data pelatihan  

Semakin kompleks suatu model, biasanya bias menurun tetapi variansi meningkat.

---

## 4.10 Regresi Ridge
Regresi Ridge merupakan metode regularisasi regresi linear yang menambahkan penalti terhadap besar bobot model, dengan tujuan menjaga kompleksitas model tetap rendah.

image.png

---

## 4.11 Regresi Lasso
Regresi Lasso adalah metode regularisasi yang mampu mengeliminasi fitur yang dianggap tidak penting dengan mendorong bobot tertentu menjadi nol.

image.png

---

## 4.12 Elastic Net
Elastic Net merupakan kombinasi dari Regresi Ridge dan Regresi Lasso, sehingga menjadi solusi tengah (*middle ground*) antara kedua metode tersebut.

image.png

---

## 4.13 Early Stopping
Early stopping merupakan teknik regularisasi yang menghentikan proses training ketika error validasi mencapai nilai minimum, sebelum model mengalami overfitting.

image.png

---

## 4.14 Mengestimasi Probabilitas pada Regresi Logistik
Regresi logistik digunakan untuk mengestimasi probabilitas suatu data termasuk ke dalam kelas tertentu.

Model ini menghasilkan kurva berbentuk sigmoid yang mencapai kondisi stabil maksimum setelah melewati titik minimum stabil sebelumnya.

---

## 4.15 Fungsi Pelatihan pada Regresi Logistik
Proses pelatihan regresi logistik menggunakan fungsi biaya berbasis **log loss** untuk mengoptimalkan probabilitas prediksi terhadap label sebenarnya.

image.png

---

## 4.16 Decision Boundaries
Decision boundary merupakan batas keputusan yang digunakan untuk memisahkan kelas-kelas dalam ruang fitur.

Pada klasifikasi dua kelas, batas ini umumnya berada pada probabilitas **50%**, di mana kedua kelas memiliki peluang yang sama.

---

## 4.17 Regresi Softmax
Regresi Softmax merupakan generalisasi dari regresi logistik untuk menangani banyak kelas (*multiclass*) tanpa perlu menggabungkan beberapa binary classifier.

Metode ini menggunakan fungsi eksponensial yang dinormalisasi dan menerapkan **cross-entropy loss**, yang mengukur perbedaan distribusi probabilitas antara prediksi model dan distribusi label sebenarnya.
