# Chapter 17: Representation Learning and Generative Learning Using Autoencoders and GANs

Bab ini membahas dua arsitektur populer untuk pembelajaran tanpa pengawasan (*unsupervised learning*): Autoencoder, yang fokus pada representasi data, dan GAN, yang fokus pada pembuatan data baru.

---

## 17.1 Pengenalan Autoencoders dan GAN
Keduanya adalah model generatif yang bekerja secara *unsupervised*.
* **Autoencoders**: Belajar menyalin input ke output dengan batasan tertentu untuk mendapatkan **Latent Representations** (representasi padat berdimensi rendah).
* **GAN (Generative Adversarial Networks)**: Terdiri dari dua jaringan yang bersaing:
    * **Generator**: Berusaha menciptakan data palsu yang mirip aslinya.
    * **Discriminator**: Berusaha membedakan mana data asli dan mana data palsu dari Generator.



---

## 17.2 Representasi Efisien & PCA
* **Undercomplete Autoencoder**: Dengan membatasi dimensi lapisan tersembunyi (*hidden layer*) agar lebih kecil dari input, model dipaksa mempelajari pola yang paling penting.
* **Hubungan dengan PCA**: Jika Autoencoder menggunakan aktivasi linear dan fungsi kerugian MSE, ia akan melakukan proyeksi data yang sama dengan **Principal Component Analysis (PCA)**.

---

## 17.3 Stacked Autoencoders
Autoencoder dengan banyak lapisan tersembunyi (Deep Autoencoders) dapat mempelajari fitur yang lebih rumit.
* **Arsitektur Simetris**: Biasanya jumlah unit di encoder dan decoder saling bercermin.
* **Tying Weights**: Teknik menghubungkan bobot decoder agar sama dengan *transpose* bobot encoder. Ini mengurangi jumlah parameter model hingga separuhnya dan mempercepat pelatihan.
* **Visualisasi**: Performa dapat dicek dengan membandingkan input dan rekonstruksi menggunakan fungsi `show_reconstructions()`. Dataset juga bisa divisualisasikan dalam 2D menggunakan **t-SNE**.

---

## 17.4 Unsupervised Pretraining
Autoencoder yang telah dilatih dapat digunakan sebagai titik awal untuk model *supervised*.
* Bagian encoder disalin ke jaringan baru, lalu ditambahkan lapisan output (misal: Softmax).
* Teknik ini membantu mencapai performa tinggi dengan data berlabel yang sedikit.
* **Greedy Layerwise Training**: Teknik melatih satu lapis autoencoder secara bergantian sebelum menggabungkannya menjadi satu model besar.

---

## 17.5 Variasi Autoencoder (CNN, RNN, Denoising)
* **Convolutional Autoencoders**: Menggunakan CNN untuk menangani data gambar (mengurangi dimensi spasial via konvolusi dan menambahnya kembali via *transposed convolution*).
* **Recurrent Autoencoders**: Menggunakan RNN untuk memproses data sekuensial atau deret waktu.
* **Denoising Autoencoders**: Input sengaja diberi *noise*. Model dipaksa untuk menghapus *noise* tersebut, sehingga ia belajar fitur-fitur yang lebih robust dan esensial.
* **Sparse Autoencoders**: Menambahkan penalti pada fungsi kerugian agar hanya sedikit neuron yang aktif pada satu waktu, mendorong representasi yang lebih efisien.

---

## 17.6 Variational Autoencoders (VAE)
Berbeda dengan autoencoder standar, VAE adalah model **probabilistik**.
* **Generative**: VAE tidak hanya memetakan input ke titik tetap, melainkan ke distribusi probabilitas.
* **Semantic Interpolation**: Kita dapat melakukan interpolasi antara dua gambar di level *coding* (ruang laten). Hasilnya adalah transisi gambar yang halus dan masuk akal secara semantik (misal: perubahan ekspresi wajah secara bertahap).



---

## 17.7 GAN (Generative Adversarial Networks)
Pelatihan GAN melibatkan dinamika permainan antara Generator dan Discriminator.
* **Nash Equilibrium**: Kondisi ideal di mana Discriminator tidak bisa lagi membedakan data asli dan palsu (akurasi 50%).
* **Mode Collapse**: Masalah di mana Generator hanya menghasilkan satu jenis data (variasi sangat rendah) karena merasa sukses menipu Discriminator dengan data tersebut.

---

## 17.8 Deep Convolutional GAN (DCGAN) & StyleGAN
* **DCGAN**: Evolusi GAN untuk gambar yang menggunakan:
    * *Strided convolutions* pada Discriminator.
    * *Transposed convolutions* pada Generator.
    * Batch Normalization (BN) dan aktivasi Leaky ReLU.
* **Progressive Growing GAN**: Melatih gambar dari resolusi kecil ($4 \times 4$) lalu perlahan menambah lapisan konvolusi hingga mencapai resolusi tinggi.
* **StyleGAN**: Menggunakan **Mapping Network** (MLP) untuk memisahkan konten gambar dari "gaya" (*style*). Menggunakan **Mixing Regularization** untuk menggabungkan gaya dari dua *coding* yang berbeda ke dalam satu gambar.
