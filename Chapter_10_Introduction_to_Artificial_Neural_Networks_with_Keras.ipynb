# Chapter 10: Artificial Neural Networks (ANN)

Artificial Neural Networks (ANN) merupakan model komputasi yang terinspirasi dari cara kerja neuron biologis dalam memproses informasi.

---

## 10.1 Asal Muasal ANN
ANN pertama kali diperkenalkan oleh ahli neurofisiologi **Warren McCulloch** dan matematikawan **Walter Pitts**. Mereka menjelaskan bagaimana neuron biologis dapat melakukan perhitungan logis dan komputasi kompleks menggunakan model matematika sederhana.

---

## 10.2 Komputasi Logis dengan Neuron
Neuron buatan menerima satu atau lebih input biner dan menghasilkan satu output biner.

Neuron akan aktif jika jumlah input yang aktif melebihi ambang batas tertentu. Dengan mekanisme ini, neuron buatan dapat merepresentasikan operasi logika seperti:
- AND
- OR
- NOT

---

## 10.3 Perceptron
Perceptron adalah salah satu arsitektur ANN paling sederhana yang didasarkan pada **Threshold Logic Unit (TLU)** atau **Linear Threshold Unit (LTU)**.

Karakteristik:
- Setiap koneksi memiliki bobot
- Input dikalikan bobot lalu dijumlahkan
- Hasilnya dilewatkan ke fungsi aktivasi **step function**
- Terdiri dari satu lapisan TLU yang terhubung langsung ke input

Perceptron dilatih menggunakan algoritma pembelajaran berbasis error untuk menyesuaikan bobot agar error minimum.

**Keterbatasan**:
Perceptron tidak mampu menyelesaikan masalah non-linear seperti **XOR**, yang kemudian diatasi oleh Multilayer Perceptron (MLP).

---

## 10.4 Multilayer Perceptron (MLP) dan Backpropagation
MLP terdiri dari:
- Satu lapisan input
- Satu atau lebih lapisan tersembunyi
- Satu lapisan output

Jika lapisan tersembunyi berjumlah banyak, jaringan ini disebut **Deep Neural Network (DNN)**.

### Backpropagation
Backpropagation adalah algoritma untuk menghitung gradien error terhadap setiap parameter jaringan.

Agar diferensial dapat dihitung, fungsi step digantikan dengan fungsi aktivasi kontinu seperti:
- Sigmoid
- Hyperbolic Tangent (tanh)
- ReLU (Rectified Linear Unit)

---

## 10.5 Regresi pada MLP
Pada regresi, lapisan output ditentukan berdasarkan jumlah nilai yang ingin diprediksi.

Karakteristik:
- Biasanya menggunakan satu neuron output
- Tidak menggunakan activation function atau menggunakan fungsi linear
- Output dapat dipastikan bernilai positif menggunakan ReLU atau Softplus
- Loss function umum: **Mean Squared Error (MSE)**

---

## 10.6 Klasifikasi Menggunakan MLP
MLP dapat digunakan untuk:
- Klasifikasi biner
- Multiclass classification
- Multilabel classification

Lapisan output biasanya menggunakan:
- Sigmoid untuk klasifikasi biner
- Softmax untuk klasifikasi multikelas

---

## 10.7 MLP Menggunakan Keras
**Keras** adalah API deep learning tingkat tinggi yang memudahkan pembangunan, pelatihan, dan evaluasi neural network.

Keras dapat digunakan bersama backend seperti:
- TensorFlow
- CNTK
- Theano

---

## 10.8 Membangun Image Classifier dengan Keras
Keras menyediakan dataset bawaan seperti **Fashion MNIST**.

Langkah umum:
1. Menggunakan **Sequential model**
2. Lapisan **Flatten** untuk mengubah gambar menjadi vektor 1D
3. Menambahkan beberapa hidden layer dengan aktivasi ReLU
4. Menentukan lapisan output
5. Melakukan **compile()** untuk menetapkan loss function dan optimizer
6. Melatih model menggunakan **fit()**
7. Mengevaluasi model dan memantau **learning curves** untuk mendeteksi underfitting atau overfitting
8. Melakukan tuning parameter jika diperlukan

---

## 10.9 Membangun Regresi dengan MLP
Untuk regresi:
- Menggunakan **Sequential API**
- Lapisan output terdiri dari satu neuron
- Tidak menggunakan activation function
- Loss function: **Mean Squared Error (MSE)**

---

## 10.10 Membangun Model Kompleks
Contoh model non-sekuensial adalah **Wide & Deep Network**.

Karakteristik:
- Input dapat langsung terhubung ke output
- Jaringan belajar pola kompleks (deep) dan aturan sederhana (wide)
- Menggunakan **Functional API**
- Setiap output memerlukan loss function masing-masing

---

## 10.11 Membangun Model Dinamis
Model dengan struktur dinamis (loop, kondisi, atau bentuk variatif) dibuat menggunakan **Subclassing API**.

Pendekatan ini:
- Mendefinisikan subclass dari `tf.keras.Model`
- Menentukan layer dalam constructor
- Mendefinisikan logika forward pass secara manual

---

## 10.12 Callbacks
Callbacks memungkinkan eksekusi fungsi tertentu pada:
- Awal atau akhir training
- Awal atau akhir setiap epoch

Contoh penggunaan:
- Early stopping
- Model checkpoint
- Learning rate scheduling

---

## 10.13 Visualisasi Menggunakan TensorBoard
TensorBoard digunakan untuk:
- Melihat learning curves
- Visualisasi grafik komputasi
- Menampilkan gambar atau embedding yang dihasilkan model

---

## 10.14 Fine-Tuning Hyperparameters
Neural network memiliki banyak hyperparameter yang perlu disesuaikan, seperti:
- Learning rate
- Jumlah layer
- Jumlah neuron
- Batch size

Metode tuning:
- Grid search
- Random search
- Cross-validation

Library Python yang umum digunakan:
- Hyperopt
- Keras Tuner
- Spearmint

---

## 10.15 Jumlah Hidden Layers
Sebagai panduan awal:
- Satu hidden layer sering kali sudah cukup
- Jumlah neuron disesuaikan dengan ukuran input dan output

Untuk dataset seperti **MNIST**, konfigurasi umum adalah:
- 3 hidden layers
- Jumlah neuron: 300, 200, dan 100
