# Chapter 12: Custom Models and Training with TensorFlow

Bab ini membahas fleksibilitas TensorFlow untuk membangun komponen *deep learning* dari nol, mulai dari operasi tensor dasar hingga pembuatan *training loop* kustom yang memberikan kontrol penuh atas model.

---

## 12.1 Pengenalan TensorFlow
TensorFlow adalah pustaka komputasi numerik yang sangat kuat, ideal untuk *machine learning* skala besar.

**Karakteristik Utama:**
* **Mirip NumPy**: Memiliki API yang serupa dengan NumPy namun dengan dukungan GPU/TPU.
* **Komputasi Terdistribusi**: Mendukung pelatihan model di banyak *node* atau perangkat.
* **JIT Compiler (XLA)**: Memiliki *Just-In-Time compiler* untuk optimasi kecepatan dan efisiensi memori.
* **Portabilitas**: Grafik komputasi dapat diekspor dan dijalankan di berbagai platform (Windows, Linux, macOS, iOS, Android).
* **Autodiff**: Mengimplementasikan diferensiasi otomatis untuk menghitung gradien secara efisien.

---

## 12.2 Operasi Tensor
Tensor adalah unit dasar data dalam TensorFlow, mirip dengan *array* pada NumPy.

**Aturan dan Fungsi:**
* **tf.constant()**: Digunakan untuk membuat tensor yang nilainya tetap (*immutable*).
* **Interoperabilitas NumPy**: TensorFlow dapat menerima NumPy *array*, namun perlu diingat bahwa TensorFlow **tidak melakukan konversi tipe data (type casting) secara otomatis** (misal: int32 ke float32 harus dilakukan manual via `tf.cast`).
* **tf.Variable**: Digunakan untuk data yang perlu dimodifikasi selama pelatihan (seperti bobot model) menggunakan metode `.assign()`.

---

## 12.3 Struktur Data Lainnya
Selain tensor standar, TensorFlow mendukung berbagai struktur data khusus:
* **Sparse Tensors (`tf.SparseTensor`)**: Efisien untuk menyimpan tensor yang sebagian besar berisi nol.
* **Ragged Tensors (`tf.RaggedTensor`)**: Untuk daftar *array* dengan panjang yang berbeda-beda.
* **String Tensors (`tf.string`)**: Untuk data tekstual (biasanya dalam format byte).
* **Tensor Arrays (`tf.TensorArray`)**: Daftar tensor yang bisa tumbuh secara dinamis.
* **Sets & Queues**: Dukungan untuk operasi himpunan dan antrean data.

---

## 12.4 Model dengan Custom Components
Keras memungkinkan pembuatan komponen kustom dengan sangat mudah. TensorFlow memastikan komponen ini dapat disimpan dan dimuat kembali (*serialization*) dengan menyimpan nama komponen di dalam fungsi atau kelasnya.

---

## 12.5 Custom Metrics vs Custom Losses
Penting untuk membedakan antara *Loss* dan *Metrics*:
* **Loss (Kerugian)**: Digunakan oleh algoritma optimasi (seperti SGD) untuk melatih model. Harus bersifat *differentiable* (dapat diturunkan) dan gradiennya tidak boleh nol di mana-mana.
* **Metrics (Metrik)**: Digunakan untuk mengevaluasi performa model oleh manusia (misal: Akurasi). Metrik lebih mudah diinterpretasikan dan tidak harus *differentiable*.

---

## 12.6 Custom Layers
Jika arsitektur model memiliki pola berulang (seperti *Inception block* atau *Residual block*), kita dapat membungkusnya menjadi **Custom Layer**. Ini membantu mempersingkat definisi model dan meningkatkan keterbacaan kode.

---

## 12.7 Custom Models
Kita dapat membuat model yang sangat kompleks dengan melakukan *subclassing* pada kelas `tf.keras.Model`.
* **Contoh**: Sebuah model dengan *residual block* yang dilewati sebanyak tiga kali sebelum masuk ke bagian output. Dengan *subclassing*, kita bisa mendefinisikan logika *forward pass* secara imperatif di dalam metode `call()`.

---

## 12.8 Customizing Losses and Metrics
Dalam skenario tingkat lanjut, *Loss* dan *Metrics* dapat ditentukan berdasarkan parameter lain di luar label (*ground truth*) dan prediksi, misalnya berdasarkan beban internal dari lapisan tertentu (seperti pada *Reconstruction Loss*).

---

## 12.9 Autodiff untuk Komputasi Gradien
TensorFlow menggunakan **`tf.GradientTape`** untuk merekam operasi otomatis guna menghitung gradien.
* Gunakan `with tf.GradientTape() as tape:` untuk membungkus operasi yang ingin dihitung gradiennya.
* Jika ingin memanggil metode `.gradient()` lebih dari satu kali, gunakan parameter `persistent=True` pada *tape* agar sumber daya tidak langsung dihapus.

---

## 12.10 Custom Training Loops
Alih-alih menggunakan `.fit()`, kita bisa membuat *loop* manual untuk kontrol total:
1.  Melakukan iterasi pada *dataset* per *batch*.
2.  Melakukan prediksi dan menghitung *loss* di dalam `GradientTape`.
3.  Menghitung gradien berdasarkan variabel yang dapat dilatih (*trainable variables*).
4.  Menggunakan *optimizer* untuk memperbarui bobot berdasarkan gradien tersebut.

---

## 12.11 AutoGraph dan Tracing
TensorFlow dapat mengonversi kode Python biasa menjadi grafik TensorFlow yang sangat cepat.
* **AutoGraph**: Menganalisis sintaks Python (seperti `for`, `if`, `while`) dan mengubahnya menjadi operasi grafik TensorFlow yang setara.
* **Tracing**: Saat fungsi yang didekorasi dengan `@tf.function` dipanggil, TensorFlow membuat grafik simbolis berdasarkan input pertama yang diterima untuk optimasi eksekusi di masa mendatang.
