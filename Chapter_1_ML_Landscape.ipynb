# Chapter 1: The ML Landscape

Machine Learning (ML) adalah bidang studi yang memberikan komputer kemampuan untuk belajar dari data tanpa diprogram secara eksplisit (Arthur Samuel, 1959).

Sebuah program komputer dikatakan *belajar* dari pengalaman **E** terhadap tugas **T** dan ukuran performa **P**, apabila performanya pada tugas **T**, yang diukur dengan **P**, meningkat seiring bertambahnya pengalaman **E** (Tom Mitchell, 1997).

Sebagai contoh, algoritma dapat dibuat untuk memilah email spam berdasarkan pola tertentu. Program tersebut diuji dan diperbaiki berulang kali hingga bekerja dengan baik. Namun, ketika variasi spam semakin beragam, pendekatan berbasis aturan menjadi sulit untuk terus diperbarui secara manual. Di sinilah Machine Learning berperan: memahami permasalahan kompleks dengan data yang sangat besar, beradaptasi terhadap data baru, serta menyelesaikan masalah yang sulit ditangani oleh algoritma tradisional.

Secara umum, algoritma Machine Learning bekerja dalam sebuah siklus: mempelajari data (*training*), mengevaluasi performa (*evaluation*), diluncurkan ke lingkungan nyata (*deployment*), menerima data terbaru, lalu kembali dipelajari. Oleh karena itu, algoritma yang baik harus melalui proses *testing* dan *validation* menggunakan berbagai *training set* yang sesuai.

---

## Jenis-Jenis Machine Learning

### 1. Supervised Learning
Menggunakan *training set* yang memiliki label (hasil yang diinginkan).  
Contoh penerapan: klasifikasi dan regresi menggunakan fitur tertentu.  
Algoritma umum: k-Nearest Neighbors (k-NN), Linear Regression, Support Vector Machines (SVM).

### 2. Unsupervised Learning
Menggunakan *training set* tanpa label.  
Contoh penerapan:  
- *Clustering* (Hierarchical Clustering Analysis / HCA)  
- *Dimensionality reduction* dan visualisasi (misalnya LLE)  
Pendekatan ini sering digunakan untuk menemukan pola tersembunyi, memisahkan data seperti gambar berdasarkan kemiripan, serta menyederhanakan data melalui ekstraksi fitur.

### 3. Semi-Supervised Learning
Menggunakan data dengan label parsial.  
Pendekatan ini umum digunakan pada sistem berskala besar, seperti pengelompokan foto otomatis (contoh: Google Photos) dan Deep Belief Networks (DBNs).

### 4. Reinforcement Learning
Sistem pembelajaran disebut *agen*, yang berinteraksi dengan lingkungan melalui mekanisme *trial and error*. Agen memperoleh *reward* ketika mengambil keputusan yang benar dan *penalty* ketika mengambil keputusan yang salah.

---

## Pendekatan Pembelajaran

### Batch Learning
Pembelajaran dilakukan sekaligus menggunakan seluruh dataset, biasanya secara *offline*. Setelah diluncurkan, model tidak dapat mempelajari data baru tanpa pelatihan ulang.

### Online Learning
Sistem mempelajari data secara bertahap dan sekuensial. Biasanya melibatkan parameter seperti *learning rate* untuk mengatur seberapa cepat sistem beradaptasi dengan data baru.

### Instance-Based Learning
Model menyimpan data pelatihan dan melakukan prediksi berdasarkan pengukuran kemiripan (*similarity*) antar data. Pendekatan ini menekankan generalisasi langsung dari contoh yang ada.

### Model-Based Learning
Model dibangun dengan memilih representasi matematis tertentu (misalnya model linear) untuk merepresentasikan data. Model ini kemudian digunakan untuk melakukan generalisasi dan prediksi, dengan parameter yang dioptimalkan agar performanya maksimal.

---

## Tantangan dalam Machine Learning

- Membutuhkan data dalam jumlah besar untuk melatih algoritma secara optimal  
- Memerlukan penyesuaian terhadap karakteristik lingkungan data (kurang general jika data berubah drastis)  
- Kualitas dataset pelatihan yang buruk, seperti adanya *noise* atau data tidak relevan  
- Banyaknya fitur yang tidak diperlukan sehingga menyulitkan proses *feature selection*  
- Risiko *overfitting* (model terlalu menyesuaikan satu dataset) dan *underfitting* (model gagal menangkap pola data dengan baik)
