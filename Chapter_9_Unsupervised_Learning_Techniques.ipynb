# Chapter 9: Unsupervised Learning Techniques

Unsupervised Learning merupakan pendekatan pembelajaran mesin yang bekerja tanpa label. Algoritma berusaha menemukan pola, struktur, atau hubungan tersembunyi dalam data.

---

## 9.1 Clustering
Clustering bertujuan untuk mengidentifikasi objek atau contoh yang memiliki kemiripan dan mengelompokkannya ke dalam satu cluster.

Teknik ini menghasilkan output tanpa label dan banyak digunakan untuk:
- Segmentasi pelanggan
- Analisis data eksploratif
- Reduksi dimensi
- Search engine
- Pengolahan citra

---

## 9.2 K-Means
K-Means digunakan ketika dataset memiliki beberapa kelompok alami.

Algoritma ini mengelompokkan setiap objek ke centroid terdekat berdasarkan jarak. Namun, K-Means cenderung kurang optimal jika diameter antar cluster sangat berbeda.

- **Hard clustering**: setiap objek hanya masuk ke satu cluster
- **Soft clustering**: objek memiliki nilai probabilitas pada setiap cluster

---

## 9.3 Metode K-Means
K-Means menggunakan metrik yang disebut **inertia**, yaitu rata-rata kuadrat jarak antara setiap objek dan centroid terdekat.

Karakteristik penting:
- Secara default mengabaikan perhitungan jarak yang tidak diperlukan untuk efisiensi
- Dapat diimplementasikan dengan **Mini-Batch K-Means** untuk dataset besar
- Nilai **k** yang optimal dapat ditentukan menggunakan **Silhouette Score**, yang memberikan koefisien siluet untuk setiap objek

---

## 9.4 Limitasi K-Means
Beberapa keterbatasan K-Means antara lain:
- Perlu dijalankan beberapa kali untuk hasil yang stabil
- Harus menentukan jumlah cluster di awal
- Kurang efektif untuk data dengan bentuk cluster yang kompleks atau tidak konveks

---

## 9.5 Clustering untuk Segmentasi Gambar
Clustering dapat digunakan dalam pengolahan citra, khususnya untuk:
- **Semantic segmentation**: membagi gambar berdasarkan jenis objek
- **Color segmentation**: membagi gambar berdasarkan kemiripan warna piksel

Teknik ini membantu memisahkan area visual tanpa label manual.

---

## 9.6 Clustering untuk Preprocessing
Clustering sering digunakan sebagai langkah preprocessing untuk mereduksi dimensi atau menyederhanakan data.

Nilai **k** dipilih berdasarkan hasil **cross-validation** yang menghasilkan performa klasifikasi terbaik pada tahap berikutnya.

---

## 9.7 Clustering untuk Semi-Supervised Learning
Clustering dapat membantu membagi data menjadi kelompok yang:
- Sudah memiliki label
- Belum memiliki label

Pendekatan ini berguna ketika hanya sebagian kecil data yang diberi label.

---

## 9.8 Active Learning
Active Learning adalah proses di mana model dilatih menggunakan data berlabel untuk memprediksi data tanpa label.

Jika model tidak yakin terhadap suatu prediksi, data tersebut diberikan kepada ahli (human-in-the-loop) untuk dilabeli hingga model mampu memberi label dengan tingkat kepercayaan tinggi.

---

## 9.9 DBSCAN
**DBSCAN (Density-Based Spatial Clustering of Applications with Noise)** mendefinisikan cluster sebagai daerah dengan kepadatan tinggi.

Karakteristik:
- Titik dengan kepadatan tinggi disebut **core instance**
- Titik di luar kepadatan dianggap **noise atau anomali**
- Tidak memerlukan jumlah cluster di awal
- Tidak dapat memprediksi cluster untuk data baru secara langsung

---

## 9.10 Gaussian Mixture Models (GMM)
Gaussian Mixture Models adalah model probabilistik yang mengasumsikan data dihasilkan dari gabungan beberapa distribusi Gaussian tanpa parameter yang diketahui.

Model ini menggunakan algoritma **Expectation-Maximization (EM)** untuk:
1. Memberikan probabilitas keanggotaan cluster
2. Memperbarui parameter distribusi

GMM bersifat **generative**, sehingga dapat menghasilkan data baru.

## 9.11 Mendeteksi Anomali
Objek yang berada di daerah dengan kepadatan rendah dapat dianggap sebagai anomali, tergantung pada threshold yang digunakan oleh model.

Pendekatan ini umum digunakan pada deteksi fraud dan monitoring sistem.

---

## 9.12 Menentukan Jumlah Cluster
Jumlah cluster optimal dapat ditentukan menggunakan kriteria informasi, seperti:
- **Bayesian Information Criterion (BIC)**
- **Akaike Information Criterion (AIC)**

Keduanya menyeimbangkan kecocokan model dan kompleksitas parameter.

---

## 9.13 Likelihood Function
- **Probability**: seberapa besar kemungkinan suatu output *x* terjadi
- **Likelihood**: seberapa besar kemungkinan parameter tertentu menghasilkan data yang diamati

Konsep likelihood sangat penting dalam model probabilistik.

---

## 9.14 Bayesian Gaussian Mixture Models
Bayesian GMM memperlakukan parameter sebagai variabel acak dan memberikan bobot yang lebih fleksibel pada setiap cluster.

Keunggulan:
- Dapat menentukan jumlah cluster secara otomatis
- Cocok untuk cluster berbentuk elips
- Lebih stabil terhadap overfitting
